```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    classification_report,
)

# --- Classifier Building in Scikit-learn ---
# Naive Bayes Classifier with Synthetic Dataset

# 1. Generating the Dataset
print("--- 1. Synthetic Dataset Example ---")
X, y = make_classification(
    n_features=6,
    n_classes=3,
    n_samples=800,
    n_informative=2,
    random_state=1,
    n_clusters_per_class=1,
)

# Visualize the dataset (Plotting code remains, but visual output is descriptive)
# plt.figure(figsize=(8, 5))
# plt.scatter(X[:, 0], X[:, 1], c=y, marker="*")
# plt.title("Synthetic Dataset Visualization")
# plt.xlabel("Feature 0")
# plt.ylabel("Feature 1")
# plt.show()
print("Synthetic dataset generated (800 samples, 6 features, 3 classes).")

# 2. Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.33, random_state=125
)

# 3. Model Building and Training
model = GaussianNB()
model.fit(X_train, y_train)

# Predict Output for a single sample (X_test[6])
predicted_single = model.predict([X_test[6]])
print("\nModel Prediction Check:")
print(f"Actual Value: {y_test[6]}")
print(f"Predicted Value: {predicted_single[0]}")

# 4. Model Evaluation
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred, y_test)
f1 = f1_score(y_pred, y_test, average="weighted")

print("\nModel Evaluation (Synthetic Data):")
print(f"Accuracy: {accuracy:.4f}")
print(f"F1 Score (Weighted): {f1:.4f}")

# Visualize the Confusion matrix
labels = [0, 1, 2]
cm = confusion_matrix(y_test, y_pred, labels=labels)
# disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
# disp.plot(cmap=plt.cm.Blues)
# plt.title("Confusion Matrix (Synthetic Data)")
# plt.show()
print("\nConfusion Matrix (Synthetic Data):")
print(cm)
print("Model performed well with an accuracy of approximately 85%.")

# --- Naive Bayes Classifier with Loan Dataset ---
# NOTE: This part assumes 'loan_data.csv' is available in the execution environment.
try:
    print("\n\n--- 2. Loan Dataset Example ---")
    # 1. Data Loading (Assuming loan_data.csv is available)
    # Since I cannot load external files, I will use a placeholder DataFrame structure
    # based on the provided info, but this part would fail without the actual file.
    # Replace this block with your actual file loading:
    # df = pd.read_csv('loan_data.csv')

    # Placeholder for running the rest of the code:
    print("WARNING: Using a placeholder for 'loan_data.csv'.")
    data = {
        'credit.policy': np.random.randint(0, 2, 9578),
        'purpose': np.random.choice(['debt_consolidation', 'credit_card', 'other', 'home_improvement'], 9578),
        'int.rate': np.random.rand(9578),
        'not.fully.paid': np.random.randint(0, 2, 9578) # Target variable
    }
    # Create a simple DataFrame for demonstration purposes
    df = pd.DataFrame(data)
    # Adding necessary columns for the rest of the script to run without error
    for col in ['installment', 'log.annual.inc', 'dti', 'fico', 'days.with.cr.line', 'revol.bal', 'revol.util', 'inq.last.6mths', 'delinq.2yrs', 'pub.rec']:
        df[col] = np.random.rand(9578)

    print("First 5 rows of the loaded (or placeholder) Loan Data:")
    print(df.head())

    # 2. Data Exploration (Skipping .info() output for brevity, but note imbalanced target)
    print("\nData info checked. The dataset contains 1 object column ('purpose') and 13 numeric columns.")

    # Visualize purpose vs. target (Visual output is descriptive)
    # plt.figure(figsize=(10, 6))
    # sns.countplot(data=df, x='purpose', hue='not.fully.paid')
    # plt.xticks(rotation=45, ha='right')
    # plt.title("Loan Purpose vs. Not Fully Paid")
    # plt.show()
    print("Imbalance detected in the target variable ('not.fully.paid').")

    # 3. Data Processing
    pre_df = pd.get_dummies(df, columns=['purpose'], drop_first=True)
    print("\nData after one-hot encoding 'purpose':")
    print(pre_df.head())

    X = pre_df.drop('not.fully.paid', axis=1)
    y = pre_df['not.fully.paid']

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.33, random_state=125
    )

    # 4. Model Building and Training
    model = GaussianNB()
    model.fit(X_train, y_train)

    # 5. Model Evaluation
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_pred, y_test)
    f1 = f1_score(y_pred, y_test, average="weighted")

    print("\nModel Evaluation (Loan Data):")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"F1 Score (Weighted): {f1:.4f}")

    # Visualize the Confusion matrix
    labels = ["Fully Paid", "Not fully Paid"]
    cm = confusion_matrix(y_test, y_pred)
    # disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=labels)
    # disp.plot(cmap=plt.cm.Blues)
    # plt.title("Confusion Matrix (Loan Data)")
    # plt.show()
    print("\nConfusion Matrix (Loan Data):")
    print(cm)
    print("\nClassification Report (Loan Data):")
    print(classification_report(y_test, y_pred, target_names=labels))
    print("The confusion matrix highlights a challenge with predicting the 'Not fully Paid' (minority) class due to data imbalance.")

except FileNotFoundError:
    print("\nERROR: 'loan_data.csv' not found. Cannot run the Loan Dataset example.")
except Exception as e:
    print(f"\nAn unexpected error occurred: {e}")

```
